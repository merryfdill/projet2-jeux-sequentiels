```python
from src.jeux import ultimatum

# Cas 1 : le répondant accepte en cas d'indifférence
resultat_acceptation = ultimatum(total=10, accepte_indifference=True)

# Cas 2 : le répondant refuse en cas d'indifférence
resultat_refus = ultimatum(total=10, accepte_indifference=False)

# Cas 3 : les deux conventions sont étudiées
resultat_deux_cas = ultimatum(total=10, accepte_indifference=None)

print("Convention 1 : acceptation en cas d'indifférence")
print(resultat_acceptation)

print("\nConvention 2 : refus en cas d'indifférence")
print(resultat_refus)

print("\nComparaison des deux conventions")
print(resultat_deux_cas)
```

L'expérience montre que le résultat du jeu de l'ultimatum dépend de la convention adoptée lorsque le répondant est indifférent entre accepter et refuser une offre nulle.

Lorsque le répondant accepte en cas d'indifférence, le proposant offre 0 et conserve la totalité du montant. Lorsqu'il refuse en cas d'indifférence, le proposant offre 1 afin de garantir l'acceptation et conserve 9 unités.

La fonction ultimatum() reproduit correctement ces deux résultats et permet de conserver les deux solutions lorsque les conventions sont étudiées simultanément.

Expérience – Conservation des plans optimaux en cas d'égalité

Un arbre de jeu a été construit avec deux actions possibles pour le joueur 1, donnant toutes les deux un gain de 5. La fonction backward_induction_tous() retourne les deux plans optimaux : Gauche et Droite.

Cette expérience confirme que l'algorithme conserve toutes les actions maximisant le gain du joueur qui décide, même lorsque les gains de l'autre joueur sont différents.

```python
from src.jeux import sous_jeu_simultane

A = [
    [2, 0],
    [0, 1]
]

B = [
    [1, 0],
    [0, 2]
]

resultat_nash = sous_jeu_simultane(A, B)

print("Équilibres de Nash :")
print(resultat_nash)
```

Expérience – Résolution d'un sous-jeu simultané

Une matrice de gains a été utilisée pour vérifier la fonction sous_jeu_simultane(). L'algorithme a identifié trois équilibres de Nash : deux équilibres purs et un équilibre mixte.

L'équilibre mixte correspond aux probabilités suivantes :
- Joueur 1 : \(P(Gauche)=2/3\), \(P(Droite)=1/3\).
- Joueur 2 : \(P(Gauche)=1/3\), \(P(Droite)=2/3\).
Les résultats obtenus sont conformes aux équilibres théoriques de cette matrice.
